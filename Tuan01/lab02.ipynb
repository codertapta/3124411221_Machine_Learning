Bài 2:Tạo các Tensor thường dùng

```python
import torch
#Tạo tensor ngẫu nhiên với 3 hàng và 4 cột
#rand không tạo ra số âm nhưng randn có thể
a=torch.rand(3,4)
e=torch.randn(3,4)
#Tạo tensor với giá trị 0 với 2 hàng và 3 cột
b=torch.zeros(2,3)
#Tạo tensor với giá trị 1 với 2 hàng và 3 cột
c=torch.ones(2,3)
#Tạo tensor là ma trận đơn vị với 3 hàng và 3 cột
d=torch.eye(3,3)
f=torch.full((5,5),10)
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

Với a và e là tạo tensor ngẫu nhiên với kích thước được cho trong ngoặc.Trong đó a không thể tạo ra số âm nhưng e có thể.Zeros và ones là tạo với giá trị toàn 0 và 1. Còn eye là ma trận đơn vị.
