Bài 6:Phép toán và nhân ma trận

```python
import torch
a=torch.randn(2,3)
b=torch.randn(2,3)
print("Tensor a:")
print(a)
print("Tensor b:")
print(b)
print("Tensor a+b:")
print(a+b)
print("Tensor a-b:")
print(a-b)
print("Tensor a*b:")
print(a*b)
print("Tensor a/b:")
print(a/b)
print("Tensor a^2:")
print(a**2)
print("Tensor a@b:")
print(a@b.T)
```

Yêu cầu:
• Phân biệt phép toán a * b với a @ b.T.:
a*b là nhân từng vị trí, còn a@b là nhân ma trận
• Cho biết shape của kết quả a @ b.T.
